```python
from langchain.chat_models import ChatOpenAI
from langchain.prompts import PromptTemplate, ChatPromptTemplate
from langchain.callbacks import StreamingStdOutCallbackHandler
import os
from dotenv import dotenv_values


env_vars = dotenv_values('.env')
os.environ['OPENAI_API_KEY'] = env_vars.get('OPENAI_API_KEY')


chat = ChatOpenAI(model="gpt-3.5-turbo",
                  temperature=0.1,
                  streaming=True,
                  callbacks=[StreamingStdOutCallbackHandler()])

poet_prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a Haikus writer. You create gorgeous Haikus, Japanese poem of seventeen syllables, \
        about programming languages with common used words.'),
    ('human', 'I want to Haikus about {p_lang}.')
])

poet_chain = poet_prompt | chat

explainer_prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a person who can explain Haikus for the public. \
        Give an explanation for the given Haikus, easy to understand in English.'),
     ('human', '{haikus}')
])

explainer_chain = explainer_prompt | chat

final_chain = {'haikus': poet_chain} | explainer_chain

final_chain.invoke({'p_lang': 'Python'})
```
